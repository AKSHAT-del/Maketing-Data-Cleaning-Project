## Step 7 Logical Intergrity (Time Travel)`

```python
impossible_masking_date = A.start_date > A.end_date
print(A.loc[impossible_masking_date,['campaign_id','start_date','end_date']].head(3))
```

```python
impossible_masking_date = A.start_date > A.end_date
print(A.loc[impossible_masking_date,['campaign_id','start_date','end_date']].head())
```

since in some of the data instances what we can see is that end_date is actually less then the start_date which is not possible so
to fix it we assume that "end_date should at least be 30 days after the start_date" for the above cases

```python
impossible_masking_date = A.start_date > A.end_date
print(A.loc[impossible_masking_date,['campaign_id','start_date','end_date']].head(3))


A.loc[impossible_masking_date,'end_date'] = A.loc[impossible_masking_date,'end_date'] + pd.Timedelta(days=30)

print("Fix Applied")
print(A.loc[impossible_masking_date,['campaign_id','start_date','end_date']].head(3))
```

now to start with fixing up the outlier data we will start with the spend columns & after that we are gonna progress to the other columns

```python
A.describe()
```

```python
Q1_spend = A.spend.quantile(0.25)
Q3_spend = A.spend.quantile(0.75)

IQR = Q3_spend - Q1_spend

upper_limit_spend = Q3_spend + (3*IQR)

outlier_mask_spend = A.spend > upper_limit_spend
print(A.loc[outlier_mask_spend,['campaign_id','spend']].head(3))

print('Fix Applied')
A.loc[outlier_mask_spend,['spend']] = upper_limit_spend
print(A.loc[outlier_mask_spend,['campaign_id','spend']].head(3))
```
